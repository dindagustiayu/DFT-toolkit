```python
import py3Dmol

# 1. Read the .xyz file content as a string
file_path = "ETKDG_alg/C10H6ClNO.xyz"
with open(file_path, "r") as f:
    xyz_data = f.read()

# 2. Initialize the 3D viewer
view = py3Dmol.view(width=600, height=400)

# 3. Add the XYZ model to the viewer
view.addModel(xyz_data, "xyz")

# 4. Set the visual style (e.g., ball-and-stick)
view.setStyle({"stick": {"radius": 0.15}, "sphere": {"scale": 0.3}})

# 5. Center the molecule and render
view.zoomTo()
view.show()
```

```python
import py3Dmol

file_path = "ETKDG_alg/C10H6ClNO.xyz"
with open(file_path, "r") as f:
    xyz_data = f.read()

# Initialize the viewer and load the structure
view = py3Dmol.view(width=600, height=400)
view.addModel(xyz_data, "xyz")
view.setStyle({"stick": {"radius": 0.15}, "sphere": {"scale": 0.3}})

# Parse the XYZ lines to extract symbols, indices, and coordinates
lines = xyz_data.strip().splitlines()
num_atoms = int(lines[0].strip())
atom_lines = lines[2 : 2 + num_atoms]  # Skip atom count (line 0) and comment (line 1)

for i, line in enumerate(atom_lines, start=1):  # Use start=0 for 0-based indexing
    parts = line.split()
    symbol = parts[0]
    x, y, z = float(parts[1]), float(parts[2]), float(parts[3])
    
    # Add label showing Symbol + Number (e.g., "C1", "O2")
    view.addLabel(
        f"{symbol}{i}",
        {
            "position": {"x": x, "y": y, "z": z},
            "fontSize": 12,
            "fontColor": "black",
            "backgroundColor": "white",
            "backgroundOpacity": 0.6,
            "showBackground": True,
        },
    )

view.zoomTo()
view.show()
```
